```python
import os
import pandas as pd

# Define relative path to raw dataset
DATA_PATH = os.path.join("..", "data", "raw", "IMDB Dataset.csv")

# Load raw dataset
df = pd.read_csv(DATA_PATH)
print(f"✅ Successfully loaded dataset from: {DATA_PATH}")
```

```python
# Check row and column counts
print(f"Dataset Shape: {df.shape[0]:,} rows | {df.shape[1]} columns\n")

# Preview first 5 rows
df.head()
```

```python
# Audit data types, memory, and non-null counts
print("--- Data Info ---")
df.info()

print("\n--- Missing Value Count ---")
missing_counts = df.isnull().sum()
print(missing_counts)
```

```python
# Check positive vs. negative sentiment balance
print("--- Class Distribution ---")
sentiment_counts = df['sentiment'].value_counts()
sentiment_percentages = df['sentiment'].value_counts(normalize=True) * 100

balance_df = pd.DataFrame({
    'Count': sentiment_counts,
    'Percentage (%)': sentiment_percentages.round(2)
})
print(balance_df)
```
